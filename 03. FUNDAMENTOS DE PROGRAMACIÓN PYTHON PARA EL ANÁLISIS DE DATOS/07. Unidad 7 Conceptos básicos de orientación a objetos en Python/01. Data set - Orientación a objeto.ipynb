## Principios básicos de la POO — Abstracción y ocultamiento
**Slide 16**

```python
class SmartTV:
    def __init__(self):
        self.__estado = "apagado"  # Atributo privado
        self.__canal = 1           # Atributo privado

    def encender(self):            # Método público
        self.__estado = "encendido"
        print("TV encendida")

    def cambiar_canal(self, canal):  # Método público
        if self.__estado == "encendido":
            self.__canal = canal
            print(f"Canal cambiado a: {canal}")
        else:
            print("La TV está apagada")

tv = SmartTV()
tv.encender()
tv.cambiar_canal(5)
```

## Principios básicos de la POO — Estado y atributos
**Slide 18**

```python
class PersonajeJuego:
    def __init__(self):
        # Estados almacenados en atributos
        self.vida = 100
        self.nivel = 1
        self.posicion_x = 0
        self.posicion_y = 0
        self.nombre = "Jugador"

jugador = PersonajeJuego()
print(f"Nombre: {jugador.nombre}, Vida: {jugador.vida}, Nivel: {jugador.nivel}")
```

## Principios básicos de la POO — Comportamiento y métodos
**Slide 20**

```python
class RobotCocina:
    def __init__(self):
        self.encendido = False
        self.velocidad = 0

    def mezclar(self, velocidad):  # Método
        self.velocidad = velocidad
        print(f"Mezclando a velocidad {velocidad}")

    def picar(self):               # Método
        print("Picando ingredientes")

    def detener(self):             # Método
        self.velocidad = 0
        print("Robot detenido")

robot = RobotCocina()
robot.mezclar(3)
robot.picar()
robot.detener()
```

## Constructores e instanciación
**Slide 22**

```python
class Estudiante:
    # Constructor
    def __init__(self, nombre, edad):
        # Inicialización de atributos
        self.nombre = nombre
        self.edad = edad
        self.notas = []

    def agregar_nota(self, nota):
        self.notas.append(nota)

# Instanciación
ana = Estudiante("Ana", 15)   # Se ejecuta el constructor
juan = Estudiante("Juan", 16) # Otra instancia

ana.agregar_nota(6.5)
ana.agregar_nota(5.8)
print(f"{ana.nombre} tiene las notas: {ana.notas}")
```

## Principios básicos de la POO — ¿Cómo se relacionan entre sí?
**Slide 23**

```python
# Ejemplo integrado: Una Smart Home
class CasaInteligente:
    def __init__(self, direccion):      # Constructor
        self.__direccion = direccion    # Atributo (oculto)
        self.luces = False              # Estado
        self.temperatura = 20          # Estado

    def ajustar_temperatura(self, temp):  # Método/Comportamiento
        self.temperatura = temp
        print(f"Temperatura ajustada a {temp}°C")

casa = CasaInteligente("Av. Principal 123")
casa.ajustar_temperatura(22)
```

## El objeto string — Métodos principales de strings
**Slide 26**

```python
texto = "  Hola Mundo Python  "

# upper() y lower(): Convertir a mayúsculas/minúsculas
print(texto.upper())
print(texto.lower())

# strip(): Eliminar espacios al inicio y final
print(texto.strip())

# replace(): Reemplazar texto
print(texto.replace("Mundo", "Clase"))

# split(): Dividir el string en una lista
palabras = texto.strip().split()
print(palabras)

# join(): Unir elementos de una lista en un string
print("-".join(palabras))
```

```python
nombre = "Ana"
apellido = "González"

# Concatenación con +
nombre_completo = nombre + " " + apellido
print(nombre_completo)

# Repetición con *
print("-" * 20)

# Formateo con f-strings (la forma más moderna)
edad = 15
print(f"Hola, soy {nombre} {apellido} y tengo {edad} años.")
```

```python
texto = "Python"

# Por índice positivo: texto[0] (primer carácter)
print(texto[0])

# Por índice negativo: texto[-1] (último carácter)
print(texto[-1])

# Por slicing: texto[inicio:fin:paso]
print(texto[0:3])    # 'Pyt'
print(texto[::2])    # 'Pto'
print(texto[::-1])   # 'nohtyP' (invertido)
```
